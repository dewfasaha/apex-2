```python

import numpy as np
import openpyxl

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import warnings
from pathlib import Path
warnings.filterwarnings('ignore')

data_path = Path("output") / "ApexPlanet_Cleaned_Dataset.xlsx"
if not data_path.is_file():
    data_path = Path("..") / data_path

df = pd.read_excel(data_path, engine='openpyxl')
```

```python
# Numerical summary
display(df.describe())

# Mean
print(df.mean(numeric_only=True))

# Median
print(df.median(numeric_only=True))

# Mode
print(df.mode().head())

# Standard deviation
print(df.std(numeric_only=True))

# Variance
print(df.var(numeric_only=True))
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['Age'], bins=15)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
df.groupby('Category')['Total_Sales'].sum().plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.show()
```

```python
df['Gender'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6, 6)
)

plt.title("Customer Gender Distribution")
plt.ylabel("")

plt.show()
```

```python
sns.boxplot(x=df['Total_Sales'])

plt.title("Sales Distribution")
plt.show()
```

```python
import sqlite3
conn = sqlite3.connect("apexplanet_sales.db")

df.to_sql(
    "sales",
    conn,
    if_exists="replace",
    index=False
)

print("Dataset successfully loaded into SQLite!")
```

```python
numeric_cols = ['Age', 'Quantity', 'Unit_Price', 'Total_Sales']

correlation = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Sales Dataset')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Quantity', y='Total_Sales')

plt.title('Quantity vs Total Sales')
plt.xlabel('Quantity')
plt.ylabel('Total Sales')
plt.show()
```

```python
gender_category = pd.crosstab(
    df['Gender'],
    df['Category'],
    values=df['Total_Sales'],
    aggfunc='sum'
)

plt.figure(figsize=(9, 5))
sns.heatmap(gender_category, annot=True, fmt='.0f', cmap='Purples')

plt.title('Sales Distribution by Gender and Category')
plt.show()
```

```python
sns.pairplot(df[numeric_cols])
plt.suptitle('Pair Plot of Numerical Variables', y=1.02)
plt.show()
```
